```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "library").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "library").is_dir():
    raise FileNotFoundError("Folder library tidak ditemukan!")

sys.path.insert(0, str(PROJECT_ROOT))

from library import (
    search_data,
    filter_data,
    sort_data,
    group_data,
    calculate_stats
)


def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")


def print_separator():
    print("=" * 60)
```

```python
def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")


def show_participants(data):
    if not data:
        print("Peserta tidak ditemukan.")
        return

    print("\nDAFTAR PESERTA")
    print("=" * 70)

    for item in data:
        print(
            f"{item['id']:02} | {item['name']:<12} | "
            f"{item['ticket']:<10} | "
            f"{item['event']:<18} | "
            f"{format_rupiah(item['fee'])}"
        )


def menu_search():
    keyword = input("Masukkan nama peserta: ").strip()

    if not keyword:
        print("Keyword tidak boleh kosong.")
        return

    result = search_data(
        participants,
        keyword,
        ["name"]
    )

    show_participants(result)


def menu_filter():
    tickets = sorted({
        item["ticket"] for item in participants
    })

    print("\nJENIS TIKET")
    for index, ticket in enumerate(tickets, 1):
        print(f"{index}. {ticket}")

    choice = input("Pilih jenis tiket: ").strip()

    if not choice.isdigit() or not 1 <= int(choice) <= len(tickets):
        print("Pilihan tidak valid.")
        return

    result = filter_data(
        participants,
        "ticket",
        tickets[int(choice) - 1]
    )

    show_participants(result)


def menu_sort():
    print("\n1. Biaya tertinggi")
    print("2. Biaya terendah")

    choice = input("Pilih: ").strip()

    if choice not in ("1", "2"):
        print("Pilihan tidak valid.")
        return

    result = sort_data(
        participants,
        "fee",
        reverse=(choice == "1")
    )

    show_participants(result)


def menu_group():
    grouped = group_data(
        participants,
        "ticket"
    )

    for ticket, items in grouped.items():
        print(f"\nJENIS TIKET: {ticket}")

        for item in items:
            print(f"- {item['name']}")


def menu_statistics():
    stats = calculate_stats(
        participants,
        "fee"
    )

    grouped = group_data(
        participants,
        "ticket"
    )

    total_revenue = stats["total"]

    print("\nEVENT REGISTRATION SUMMARY")
    print("=" * 50)

    print(f"Total peserta       : {stats['count']}")
    print(f"Total pendapatan    : {format_rupiah(total_revenue)}")
    print(f"Rata-rata biaya     : {format_rupiah(stats['average'])}")
    print(f"Biaya tertinggi     : {format_rupiah(stats['maximum'])}")
    print(f"Biaya terendah      : {format_rupiah(stats['minimum'])}")

    print("\nPESERTA PER JENIS TIKET")

    for ticket, items in grouped.items():
        print(f"{ticket}: {len(items)} peserta")
```

```python
def main_menu():
    while True:
        print("\n" + "=" * 50)
        print("EVENT REGISTRATION MANAGER".center(50))
        print("=" * 50)

        print("1. Lihat seluruh peserta")
        print("2. Cari peserta")
        print("3. Filter jenis tiket")
        print("4. Urutkan biaya pendaftaran")
        print("5. Kelompokkan jenis tiket")
        print("6. Statistik pendaftaran")
        print("0. Keluar")

        choice = input("\nPilih menu: ").strip()

        if choice == "1":
            show_participants(participants)
        elif choice == "2":
            menu_search()
        elif choice == "3":
            menu_filter()
        elif choice == "4":
            menu_sort()
        elif choice == "5":
            menu_group()
        elif choice == "6":
            menu_statistics()
        elif choice == "0":
            print("Terima kasih telah menggunakan Event Registration Manager!")
            break
        else:
            print("Pilihan tidak valid.")


main_menu()
```
