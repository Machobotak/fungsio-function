```python
# Import library
from library import (
    search_data,
    filter_data,
    sort_data,
    group_data,
    calculate_stats
)

# Helper untuk menampilkan nominal Rupiah
def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")

def print_separator():
    print("=" * 60)

print("Finance Manager berhasil dijalankan!")
```

```python
def show_transactions(data):
    if not data:
        print("\nTidak ada transaksi.")
        return

    print_separator()
    print("DAFTAR TRANSAKSI".center(60))
    print_separator()

    for item in data:
        print(f"ID       : {item['id']}")
        print(f"Nama     : {item['name']}")
        print(f"Tipe     : {item['type'].upper()}")
        print(f"Kategori : {item['category']}")
        print(f"Nominal  : {format_rupiah(item['amount'])}")
        print(f"Tanggal  : {item['date']}")
        print("-" * 60)
```

```python
def menu_search():
    keyword = input("\nMasukkan kata kunci: ").strip()

    if not keyword:
        print("Kata kunci tidak boleh kosong.")
        return

    result = search_data(
        transactions,
        keyword,
        ["name", "category"]
    )

    print(f"\nDitemukan {len(result)} transaksi.")
    show_transactions(result)
```

```python
def menu_filter():
    print("\nFILTER TRANSAKSI")
    print("1. Berdasarkan jenis transaksi")
    print("2. Berdasarkan kategori")

    choice = input("Pilih menu: ").strip()

    if choice == "1":
        field = "type"
        print("\n1. Income")
        print("2. Expense")

        option = input("Pilih jenis: ").strip()

        values = {
            "1": "income",
            "2": "expense"
        }

        if option not in values:
            print("Pilihan tidak valid.")
            return

        value = values[option]

    elif choice == "2":
        field = "category"

        categories = sorted({
            item["category"] for item in transactions
        })

        print("\nKategori tersedia:")

        for index, category in enumerate(categories, 1):
            print(f"{index}. {category}")

        option = input("Pilih kategori: ").strip()

        if not option.isdigit() or not 1 <= int(option) <= len(categories):
            print("Pilihan tidak valid.")
            return

        value = categories[int(option) - 1]

    else:
        print("Pilihan tidak valid.")
        return

    result = filter_data(transactions, field, value)

    print(f"\nDitemukan {len(result)} transaksi.")
    show_transactions(result)
```

```python
def menu_sort():
    print("\nURUTKAN TRANSAKSI")
    print("1. Nominal terkecil")
    print("2. Nominal terbesar")

    choice = input("Pilih menu: ").strip()

    if choice not in ["1", "2"]:
        print("Pilihan tidak valid.")
        return

    result = sort_data(
        transactions,
        "amount",
        reverse=(choice == "2")
    )

    show_transactions(result)
```

```python
def menu_group():
    grouped = group_data(
        transactions,
        "category"
    )

    print("\nPENGELOMPOKAN TRANSAKSI")
    print_separator()

    for category, items in grouped.items():
        print(f"\nKategori: {category}")

        for item in items:
            print(
                f"  - {item['name']}: "
                f"{format_rupiah(item['amount'])}"
            )
```

```python
def menu_statistics():
    income = filter_data(
        transactions,
        "type",
        "income"
    )

    expenses = filter_data(
        transactions,
        "type",
        "expense"
    )

    income_stats = calculate_stats(
        income,
        "amount"
    )

    expense_stats = calculate_stats(
        expenses,
        "amount"
    )

    balance = (
        income_stats["total"] -
        expense_stats["total"]
    )

    print_separator()
    print("FINANCIAL SUMMARY".center(60))
    print_separator()

    print(f"Total pemasukan   : {format_rupiah(income_stats['total'])}")
    print(f"Total pengeluaran : {format_rupiah(expense_stats['total'])}")
    print(f"Saldo akhir       : {format_rupiah(balance)}")

    print_separator()

    print(f"Jumlah pemasukan  : {income_stats['count']}")
    print(f"Jumlah pengeluaran: {expense_stats['count']}")

    print(
        f"Rata-rata pengeluaran: "
        f"{format_rupiah(expense_stats['average'])}"
    )

    print_separator()
```

```python
def main_menu():
    while True:
        print("\n")
        print_separator()
        print("FINANCE MANAGER".center(60))
        print("PERSONAL FINANCE MANAGEMENT SYSTEM".center(60))
        print_separator()

        print("1. Lihat seluruh transaksi")
        print("2. Cari transaksi")
        print("3. Filter transaksi")
        print("4. Urutkan transaksi")
        print("5. Kelompokkan transaksi")
        print("6. Statistik keuangan")
        print("0. Keluar")

        print_separator()

        choice = input("Pilih menu: ").strip()

        if choice == "1":
            show_transactions(transactions)

        elif choice == "2":
            menu_search()

        elif choice == "3":
            menu_filter()

        elif choice == "4":
            menu_sort()

        elif choice == "5":
            menu_group()

        elif choice == "6":
            menu_statistics()

        elif choice == "0":
            print("\nTerima kasih telah menggunakan Finance Manager!")
            break

        else:
            print("\nPilihan tidak valid. Silakan coba lagi.")


# Menjalankan aplikasi
main_menu()
```
