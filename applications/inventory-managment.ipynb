```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "library").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "library").is_dir():
    raise FileNotFoundError("Folder library tidak ditemukan!")

sys.path.insert(0, str(PROJECT_ROOT))

from library import (
    search_data,
    filter_data,
    sort_data,
    group_data,
    calculate_stats
)


def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")


def print_separator():
    print("=" * 60)
```

```python
products = [
    {
        "id": 1,
        "name": "Laptop ASUS",
        "category": "Electronics",
        "stock": 10,
        "price": 8500000
    },
    {
        "id": 2,
        "name": "Mouse Logitech",
        "category": "Accessories",
        "stock": 25,
        "price": 150000
    },
    {
        "id": 3,
        "name": "Keyboard Mechanical",
        "category": "Accessories",
        "stock": 15,
        "price": 450000
    },
    {
        "id": 4,
        "name": "Monitor LG",
        "category": "Electronics",
        "stock": 8,
        "price": 2200000
    },
    {
        "id": 5,
        "name": "Buku Python",
        "category": "Books",
        "stock": 30,
        "price": 120000
    },
    {
        "id": 6,
        "name": "Webcam Logitech",
        "category": "Electronics",
        "stock": 12,
        "price": 750000
    },
    {
        "id": 7,
        "name": "Flashdisk 64GB",
        "category": "Accessories",
        "stock": 40,
        "price": 85000
    },
    {
        "id": 8,
        "name": "Buku Database",
        "category": "Books",
        "stock": 20,
        "price": 95000
    },
    {
        "id": 9,
        "name": "Printer Canon",
        "category": "Electronics",
        "stock": 5,
        "price": 1800000
    },
    {
        "id": 10,
        "name": "Headset",
        "category": "Accessories",
        "stock": 18,
        "price": 250000
    }
]

print(f"Total jenis barang: {len(products)}")
```

```python
def show_products(data):
    if not data:
        print("\nBarang tidak ditemukan.")
        return

    print_separator()
    print("DAFTAR INVENTARIS".center(60))
    print_separator()

    for item in data:
        print(f"ID       : {item['id']}")
        print(f"Nama     : {item['name']}")
        print(f"Kategori : {item['category']}")
        print(f"Stok     : {item['stock']} unit")
        print(f"Harga    : {format_rupiah(item['price'])}")
        print("-" * 60)
```

```python
def menu_search():
    keyword = input("\nMasukkan nama barang: ").strip()

    if not keyword:
        print("Keyword tidak boleh kosong.")
        return

    result = search_data(
        products,
        keyword,
        ["name", "category"]
    )

    print(f"\nDitemukan {len(result)} barang.")
    show_products(result)
```

```python
def menu_filter():
    print("\nFILTER INVENTARIS")
    print("1. Berdasarkan kategori")

    categories = sorted({
        item["category"] for item in products
    })

    for index, category in enumerate(categories, 1):
        print(f"{index}. {category}")

    choice = input("\nPilih kategori: ").strip()

    if not choice.isdigit():
        print("Pilihan tidak valid.")
        return

    index = int(choice) - 1

    if index < 0 or index >= len(categories):
        print("Kategori tidak ditemukan.")
        return

    result = filter_data(
        products,
        "category",
        categories[index]
    )

    show_products(result)
```

```python
def menu_sort():
    print("\nURUTKAN BARANG")
    print("1. Harga termurah")
    print("2. Harga termahal")
    print("3. Stok paling sedikit")
    print("4. Stok paling banyak")

    choice = input("Pilih menu: ").strip()

    options = {
        "1": ("price", False),
        "2": ("price", True),
        "3": ("stock", False),
        "4": ("stock", True)
    }

    if choice not in options:
        print("Pilihan tidak valid.")
        return

    field, reverse = options[choice]

    result = sort_data(
        products,
        field,
        reverse
    )

    show_products(result)
```

```python
def menu_group():
    grouped = group_data(
        products,
        "category"
    )

    print("\nPENGELOMPOKAN BARANG")
    print_separator()

    for category, items in grouped.items():
        print(f"\nKategori: {category}")

        for item in items:
            print(
                f"  - {item['name']} "
                f"({item['stock']} unit)"
            )
```

```python
def menu_statistics():
    stock_stats = calculate_stats(
        products,
        "stock"
    )

    price_stats = calculate_stats(
        products,
        "price"
    )

    inventory_value = sum(
        item["stock"] * item["price"]
        for item in products
    )

    print_separator()
    print("INVENTORY SUMMARY".center(60))
    print_separator()

    print(f"Jumlah jenis barang : {len(products)}")
    print(f"Total stok          : {stock_stats['total']} unit")
    print(f"Rata-rata stok      : {stock_stats['average']:.2f} unit")
    print(f"Harga rata-rata     : {format_rupiah(price_stats['average'])}")
    print(f"Nilai persediaan    : {format_rupiah(inventory_value)}")

    print_separator()
```

```python
def main_menu():
    while True:
        print("\n")
        print_separator()
        print("INVENTORY MANAGEMENT".center(60))
        print("INVENTORY & STOCK MANAGEMENT SYSTEM".center(60))
        print_separator()

        print("1. Lihat seluruh barang")
        print("2. Cari barang")
        print("3. Filter barang")
        print("4. Urutkan barang")
        print("5. Kelompokkan barang")
        print("6. Statistik inventaris")
        print("0. Keluar")

        print_separator()

        choice = input("Pilih menu: ").strip()

        if choice == "1":
            show_products(products)

        elif choice == "2":
            menu_search()

        elif choice == "3":
            menu_filter()

        elif choice == "4":
            menu_sort()

        elif choice == "5":
            menu_group()

        elif choice == "6":
            menu_statistics()

        elif choice == "0":
            print("\nTerima kasih telah menggunakan Inventory Management!")
            break

        else:
            print("\nPilihan tidak valid.")


main_menu()
```
