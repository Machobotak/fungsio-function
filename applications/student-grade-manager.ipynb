```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "library").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "library").is_dir():
    raise FileNotFoundError("Folder library tidak ditemukan!")

sys.path.insert(0, str(PROJECT_ROOT))

from library import (
    search_data,
    filter_data,
    sort_data,
    group_data,
    calculate_stats
)


def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")


def print_separator():
    print("=" * 60)
```

```python
students = [
    {"id": 1, "name": "Andi", "class": "IF-A", "course": "Database", "score": 85},
    {"id": 2, "name": "Budi", "class": "IF-A", "course": "Database", "score": 78},
    {"id": 3, "name": "Citra", "class": "IF-B", "course": "Database", "score": 92},
    {"id": 4, "name": "Dina", "class": "IF-B", "course": "Database", "score": 88},
    {"id": 5, "name": "Eko", "class": "IF-A", "course": "Database", "score": 65},
    {"id": 6, "name": "Fajar", "class": "IF-C", "course": "Database", "score": 90},
    {"id": 7, "name": "Gita", "class": "IF-C", "course": "Database", "score": 75},
    {"id": 8, "name": "Hana", "class": "IF-B", "course": "Database", "score": 95},
    {"id": 9, "name": "Indra", "class": "IF-A", "course": "Database", "score": 82},
    {"id": 10, "name": "Joko", "class": "IF-C", "course": "Database", "score": 70}
]
```

```python
def show_students(data):
    if not data:
        print("Data mahasiswa tidak ditemukan.")
        return

    print("\nDAFTAR NILAI MAHASISWA")
    print("=" * 65)

    for item in data:
        print(
            f"{item['id']:02} | {item['name']:<10} | "
            f"{item['class']} | {item['course']} | "
            f"{item['score']}"
        )


def menu_search():
    keyword = input("Masukkan nama mahasiswa: ").strip()

    if not keyword:
        print("Keyword tidak boleh kosong.")
        return

    result = search_data(students, keyword, ["name"])
    show_students(result)


def menu_filter():
    classes = sorted({item["class"] for item in students})

    print("\nDAFTAR KELAS")
    for index, value in enumerate(classes, 1):
        print(f"{index}. {value}")

    choice = input("Pilih kelas: ").strip()

    if not choice.isdigit() or not 1 <= int(choice) <= len(classes):
        print("Pilihan tidak valid.")
        return

    result = filter_data(
        students,
        "class",
        classes[int(choice) - 1]
    )

    show_students(result)


def menu_sort():
    print("\n1. Nilai tertinggi")
    print("2. Nilai terendah")

    choice = input("Pilih: ").strip()

    if choice not in ("1", "2"):
        print("Pilihan tidak valid.")
        return

    result = sort_data(
        students,
        "score",
        reverse=(choice == "1")
    )

    show_students(result)


def menu_group():
    grouped = group_data(students, "class")

    for class_name, items in grouped.items():
        print(f"\nKELAS {class_name}")

        for item in items:
            print(f"- {item['name']}: {item['score']}")


def menu_statistics():
    stats = calculate_stats(students, "score")

    print("\nSTUDENT GRADE SUMMARY")
    print("=" * 50)
    print(f"Jumlah mahasiswa : {stats['count']}")
    print(f"Total nilai      : {stats['total']}")
    print(f"Rata-rata nilai  : {stats['average']:.2f}")
    print(f"Nilai tertinggi  : {stats['maximum']}")
    print(f"Nilai terendah   : {stats['minimum']}")
```

```python
def main_menu():
    while True:
        print("\n" + "=" * 50)
        print("STUDENT GRADE MANAGER".center(50))
        print("=" * 50)

        print("1. Lihat seluruh nilai")
        print("2. Cari mahasiswa")
        print("3. Filter berdasarkan kelas")
        print("4. Urutkan nilai")
        print("5. Kelompokkan berdasarkan kelas")
        print("6. Statistik nilai")
        print("0. Keluar")

        choice = input("\nPilih menu: ").strip()

        if choice == "1":
            show_students(students)
        elif choice == "2":
            menu_search()
        elif choice == "3":
            menu_filter()
        elif choice == "4":
            menu_sort()
        elif choice == "5":
            menu_group()
        elif choice == "6":
            menu_statistics()
        elif choice == "0":
            print("Terima kasih telah menggunakan Student Grade Manager!")
            break
        else:
            print("Pilihan tidak valid.")


main_menu()
```
