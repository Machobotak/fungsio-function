```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "library").is_dir():
    PROJECT_ROOT = PROJECT_ROOT.parent

if not (PROJECT_ROOT / "library").is_dir():
    raise FileNotFoundError("Folder library tidak ditemukan!")

sys.path.insert(0, str(PROJECT_ROOT))

from library import (
    search_data,
    filter_data,
    sort_data,
    group_data,
    calculate_stats
)


def format_rupiah(amount):
    return f"Rp {amount:,.0f}".replace(",", ".")


def print_separator():
    print("=" * 60)
```

```python
books = [
    {"id": 1, "title": "Python Programming", "category": "Technology", "year": 2022, "copies": 10},
    {"id": 2, "title": "Database System", "category": "Technology", "year": 2021, "copies": 8},
    {"id": 3, "title": "Clean Code", "category": "Technology", "year": 2008, "copies": 5},
    {"id": 4, "title": "Atomic Habits", "category": "Self Development", "year": 2018, "copies": 12},
    {"id": 5, "title": "Deep Work", "category": "Self Development", "year": 2016, "copies": 7},
    {"id": 6, "title": "Laskar Pelangi", "category": "Novel", "year": 2005, "copies": 15},
    {"id": 7, "title": "Bumi Manusia", "category": "Novel", "year": 1980, "copies": 9},
    {"id": 8, "title": "The Pragmatic Programmer", "category": "Technology", "year": 1999, "copies": 6},
    {"id": 9, "title": "Filosofi Teras", "category": "Self Development", "year": 2018, "copies": 11},
    {"id": 10, "title": "Negeri 5 Menara", "category": "Novel", "year": 2009, "copies": 13}
]
```

```python
def show_books(data):
    if not data:
        print("Buku tidak ditemukan.")
        return

    print("\nKATALOG PERPUSTAKAAN")
    print("=" * 70)

    for book in data:
        print(
            f"{book['id']:02} | {book['title']:<30} | "
            f"{book['category']:<18} | "
            f"{book['year']} | {book['copies']} eksemplar"
        )


def menu_search():
    keyword = input("Masukkan judul atau kategori buku: ").strip()

    if not keyword:
        print("Keyword tidak boleh kosong.")
        return

    result = search_data(
        books,
        keyword,
        ["title", "category"]
    )

    show_books(result)


def menu_filter():
    categories = sorted({book["category"] for book in books})

    print("\nKATEGORI BUKU")
    for index, category in enumerate(categories, 1):
        print(f"{index}. {category}")

    choice = input("Pilih kategori: ").strip()

    if not choice.isdigit() or not 1 <= int(choice) <= len(categories):
        print("Pilihan tidak valid.")
        return

    result = filter_data(
        books,
        "category",
        categories[int(choice) - 1]
    )

    show_books(result)


def menu_sort():
    print("\n1. Tahun terbit terbaru")
    print("2. Tahun terbit terlama")

    choice = input("Pilih: ").strip()

    if choice not in ("1", "2"):
        print("Pilihan tidak valid.")
        return

    result = sort_data(
        books,
        "year",
        reverse=(choice == "1")
    )

    show_books(result)


def menu_group():
    grouped = group_data(books, "category")

    for category, items in grouped.items():
        print(f"\nKATEGORI: {category}")

        for book in items:
            print(f"- {book['title']} ({book['year']})")


def menu_statistics():
    year_stats = calculate_stats(books, "year")
    copy_stats = calculate_stats(books, "copies")

    print("\nLIBRARY COLLECTION SUMMARY")
    print("=" * 50)
    print(f"Jumlah judul buku    : {len(books)}")
    print(f"Total eksemplar      : {copy_stats['total']}")
    print(f"Rata-rata eksemplar  : {copy_stats['average']:.2f}")
    print(f"Tahun terbit tertua  : {year_stats['minimum']}")
    print(f"Tahun terbit terbaru: {year_stats['maximum']}")
```

```python
def main_menu():
    while True:
        print("\n" + "=" * 50)
        print("LIBRARY MANAGEMENT".center(50))
        print("=" * 50)

        print("1. Lihat seluruh buku")
        print("2. Cari buku")
        print("3. Filter kategori")
        print("4. Urutkan tahun terbit")
        print("5. Kelompokkan kategori")
        print("6. Statistik koleksi")
        print("0. Keluar")

        choice = input("\nPilih menu: ").strip()

        if choice == "1":
            show_books(books)
        elif choice == "2":
            menu_search()
        elif choice == "3":
            menu_filter()
        elif choice == "4":
            menu_sort()
        elif choice == "5":
            menu_group()
        elif choice == "6":
            menu_statistics()
        elif choice == "0":
            print("Terima kasih telah menggunakan Library Management!")
            break
        else:
            print("Pilihan tidak valid.")


main_menu()
```
